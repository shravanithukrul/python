```python
#Task 1: Store Train Data Using Dictionaries
```

```python
train = {
    "Train_Number": 12951,
    "Train_Name": "Mumbai Rajdhani Express",
    "Source": "Mumbai",
    "Destination": "Delhi",
    "Available_Seats": 120
}
```

```python
print("Train Information:")
print(train)

print("\nTrain Number:", train["Train_Number"])
print("Train Name:", train["Train_Name"])
print("Source Station:", train["Source"])
print("Destination Station:", train["Destination"])
print("Available Seats:", train["Available_Seats"])
```

```python
#Task 2: Access and Update Values
```

```python
#1 Create a Dictionary
```

```python
train = {
    "Train_Number": 12951,
    "Train_Name": "Mumbai Rajdhani Express",
    "Available_Seats": 120
}
```

```python
#2 Display Original Dictionary
```

```python
print(train)
```

```python
#3 Access Dictionary Values
```

```python
print("Train Name:", train["Train_Name"])
print("Available Seats:", train["Available_Seats"])
```

```python
#4 Update Seat Availability
```

```python
train["Available_Seats"] = 95
print(train)
```

```python
#5 Add New Value Using update()
```

```python
train.update({"Platform": 4})
print(train)
```

```python
#6 Remove Value Using pop()
```

```python
train.pop("Platform")
print(train)
```

```python
#7 Display Updated Dictionary
```

```python
print("\nUpdated Train Record:")
print(train)
```
